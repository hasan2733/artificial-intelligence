```python
number = int(input("Enter a number : "))
fact = 1
if(number == 0):
  print("The factorial of 0 is 1")
else:
  for i in range(1,number + 1):
    fact = fact * i
print("Factorial of number " ,number," is : ",fact)
```

```python
def factorial_cal(num):
  fact = 1
  if(num == 0):
    return fact
  else:
    return factorial_cal(num - 1) * num


num = int(input("Enter a number : "))

print("Factorial of number " ,num," is : ",factorial_cal(num))
```
