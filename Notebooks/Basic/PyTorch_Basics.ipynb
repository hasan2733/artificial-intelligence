```python
N = 10

D_in = 1
D_out = 1

x = torch.randn(N,D_in)

true_W = torch.tensor([[2.0]])
true_B = torch.tensor([1.0])

y_true = x @ true_W + true_B + torch.randn(N,D_out)*0.1
```

```python
w = torch.randn(D_in,D_out,requires_grad=True)
b = torch.randn(1,D_out,requires_grad=True)

print(w)
print(b)
```

```python
y_pred = x @ w + b
```

```python
print(y_pred)
```

```python
print(y_true)
```

```python
error = y_true - y_pred
mean_squared_error = torch.mean(error**2)
print(mean_squared_error)
```

```python
mean_squared_error.backward()
```

```python
print(w.grad)
print(b.grad)
```

```python
learning_rate,epoch = 0.1,100

w = torch.randn(D_in,D_out,requires_grad=True)
b = torch.randn(1,D_out,requires_grad=True)

for i in range(epoch):
  y_hat = x @ w + b
  error = y_true - y_hat
  mean_squared_error = torch.mean(error**2)
  mean_squared_error.backward()

  with torch.no_grad():
    w -= learning_rate*w.grad
    b -= learning_rate*b.grad
    w.grad.zero_()
    b.grad.zero_()
```

```python
import torch.nn as nn

class LinearRegression(nn.Module):
  def __init__(self, in_features,out_features):
    super().__init__()

    self.linear_layer = nn.Linear(in_features=in_features,out_features=out_features)

  def forward(self,x):
    return self.linear_layer(x)

model = LinearRegression(in_features=1,out_features=1)
```

```python
print(model)
```

```python
import torch.nn as nn

class FeedForward(nn.Module):
  def __init__(self,emmbeding_dim,ffn_dim):
    super().__init__()
    self.layer1 = nn.Linear(emmbeding_dim,ffn_dim)
    self.activation = nn.GELU()
    self.layer2 = nn.Linear(ffn_dim,emmbeding_dim)

  def forward(self,x):
    x = self.layer1(x)
    x = self.activation(x)
    x = self.layer2(x)
    return x
```
