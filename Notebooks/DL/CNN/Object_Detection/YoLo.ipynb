```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
```

```python
res = model.predict('2_modon.jpeg',imgsz = 640)
```

```python
import matplotlib.pyplot as plt
import cv2

img_bgr = res[0].plot()
plt.figure(figsize=(12,8))
img = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.axis('off')
plt.show()
```
