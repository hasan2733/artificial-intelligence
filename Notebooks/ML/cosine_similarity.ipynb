```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
newsgroup = fetch_20newsgroups(subset='train',remove=('headers','footers','quotes'))
documents = newsgroup.data
categories = newsgroup.target
target_name = newsgroup.target_names

print('Number of document ',len(documents))
print('Number of categories ',len(target_name))
```

```python
for i in range(3):
  print('$'*80)
  print('Document ',i)
  print(documents[i][:1000])
  print('Target name ',target_name[categories[i]])
```

```python
vectorizer = TfidfVectorizer(stop_words='english',max_features=10000)

tf_matrix = vectorizer.fit_transform(documents)
print('Shape of tf matrix ',tf_matrix.shape)
```

```python
query_document = tf_matrix[0]
```

```python
similarities = cosine_similarity(query_document,tf_matrix)
```

```python
similarity_score = similarities[0]
similar_indices = similarity_score.argsort()[::-1]
top_5 = similar_indices[1:6]
```

```python
print(top_5)
```

```python
print("QUERY DOCUMENT")
print("$" * 80)
print(documents[0][:1000])

print("\nCATEGORY:")
print(target_name[categories[0]])

print("\nMOST SIMILAR DOCUMENTS")
print("=" * 80)

for rank, index in enumerate(top_5, start=1):

    print(f"\nRank {rank}")
    print("Document index:", index)
    print("Cosine similarity:", similarity_score[index])
    print("Category:", target_name[categories[index]])
    print("Text:")
    print(documents[index][:500])
```
